### Математические операторы
"+" - сложение

"-" - вычитание

"/" - деление

"//" - целочисленное деление

"%" - остаток от деления

"*" - умножение

"**" - возведение в степень

````python
a = 7
b = 2
print("a + b = ", a + b)
print("a - b = ", a - b)
print("a / b = ", a / b)
print("a // b = ", a // b)
print("a % b = ", a % b) # В данном случае ближайшее число к 7, которое делится на 2 без остатка, это 6. Поэтому остаток от деления равен 7 - 6 = 1
print("a * b = ", a * b)
print("a ** b = ", a ** b)
````

### Логические операторы

"==" - Возвращает True, если оба операнда равны. Иначе возвращает False.

"!=" - Возвращает True, если оба операнда НЕ равны. Иначе возвращает False.

"> (больше чем)" - Возвращает True, если первый операнд больше второго.

"< (меньше чем)" - Возвращает True, если первый операнд меньше второго.

">= (больше или равно)" - Возвращает True, если первый операнд больше или равен второму.

"<= (меньше или равно)" - Возвращает True, если первый операнд меньше или равен второму.

"and" - логическое умножение

"or" - логическое сложение

"not" - логическое отрицание

````python
print("a > b =",a > b)
print("a < b = ",a > b)
print("a == b = ",a == b)
print("a != b = ",a != b)
print("a >= b = ",a >= b)
print("a < b and a > b = ", a < b and a > b)
print("a < b or a > b = ", a < b or a > b)
print("not a < b = ", not a < b)
````

Расчет дифференциального уравнения второго порядка для угла $\theta$ маятника, на который действует сила тяжести с трением:
$\theta''(t) + b*\theta'(t) + c*sin(\theta(t)) = 0$, $b$ и $c$ — положительные константы

Данное уравнение можно представить как систему уравнений

$\theta'(t) = \omega(t)$

$\omega'(t) = -b*\omega(t) - c*\sin(\theta(t))$

Пусть $y$ будет вектором ${\theta, \omega}, тогда решение

````python
from scipy.integrate import odeint

def pend(y, t, b, c):
    theta, omega = y
    dydt = [omega, -b*omega - c*np.sin(theta)]
    return dydt


b = 0.25
c = 5.0
y0 = [np.pi - 0.1, 0.0]
t = np.linspace(0, 10, 101)

sol = odeint(pend, y0, t, args=(b, c))

plt.plot(t, sol[:, 0], 'b', label='theta(t)')
plt.plot(t, sol[:, 1], 'g', label='omega(t)')
plt.legend(loc='best')
plt.xlabel('t')
plt.grid()
plt.show()
````

### Лямбда функции

Лямбда-выражения в языке Python представляют небольшие анонимные функции, которые определяются с помощью оператора lambda. Формальное определение лямбда-выражения: `lambda [параметры] : инструкция`

`message = lambda: print("hello")`
 
Вызов `message()` выведет сообщение hello

Здесь лямбда-выражение присваивается переменной message. Это лямбда-выражение не имеет параметров, ничего не возвращает и просто выводит строку "hello" на консоль. И через переменную message мы можем вызвать это лямбда-выражение как обычную функцию. Фактически оно аналогично следующей функции:

`def message(): `
    `print("hello")`

````python
square = lambda n: n * n
 
print(square(4))
print(square(5)) 
````

Хотя лямбда-выражения позволяют немного сократить определения функций, тем не менее они ограничены тем, что они могут выполнять только одну инструкцию. Однако они могут быть довольно удобны в тех случаях, когда необходимо использовать функцию для передачи в качестве параметра или возвращения в другой функции. 

````python
def do_operation(a, b, operation):
    result = operation(a, b)
    print(f"result = {result}")

do_operation(5, 4, lambda a, b: a + b)
do_operation(5, 4, lambda a, b: a * b)
````

### Задание 2 Числа Фибоначчи

Напишите программу, которая спрашивает пользователя, сколько чисел Фибоначчи нужно сгенерировать, а затем выводит числа ва строку Воспользуйтесь этой возможностью, чтобы подумать о том, как вы можете использовать функции. Обязательно попросите пользователя ввести количество чисел в последовательности, которую нужно сгенерировать.

( Подсказка: последовательность Фибоначчи - это последовательность чисел, где следующее число в последовательности - это сумма двух предыдущих чисел в последовательности. выглядит так: 1, 1, 2, 3, 5, 8, 13,…  формулу легко можно найти в открытых источниках)

````python
def generate_fibonacci(n):
    # Создаем пустой список для хранения чисел
    fib = []
    # Начальные значения для первых двух чисел
    a,b = 1, 1

    # Повторим цикл n раз
    for _ in range(n):
        # Добавляем текущее число в список
        fib.append(a)
        # Сдвигаем переменные для следующего шага
        a,b = b, a + b
    # Возвращаем готовый список чисел
    return fib
````

````python
# Запрашиваем у пользователя количество чисел
count = int(input("Сколько чисел Фибоначчи нужно сгенерировать?"))

# Вызываем функцию и сохраняем результат в переменную
result = generate_fibonacci(count)

# Выводим результат в одну строку через пробел
print(*result)
````

### Задание 3 Циклы

1. Создать функцию, которая: 
   1. На вход принимает n
   2. Генерирует матрицу n на n
   3. Проверяет сумму чисел в диагонали и сравнивает ее с суммой числе по строкам и столбцам:
      1. Если сумма чисел любого столбца или строки равна сумме чисел только в одной диагонали, то выводит сообщение 'Сумма одной диагонали {сумма}'
      2. Если сумма чисел любого столбца или строки равна сумме чисел в вух диагоналях, то выводит сообщение 'Сумма одной диагонали {сумма}, а второй {сумма 2}'
      3. Иначе выводит сообщение 'Сумм нет'
2. Написать функцию, которая:
   1. Запрашивает у пользователя строку с логином
   2. Проверяет строку на наличие логина в справочнике "Мавпродош, Лорнектиф, Древерол, Фиригарпиг, Клодобродыч"
   3. Выводит сообщение с информации есть логин с таким именем или нет.
   4. Функция запрашивает логин, пока пользователь не введет такой, который есть в справочнике. 
   5. Если было совпадение по логину функция завершает работу.

````python
# Задание 1

# Шаг 1: Генерация матрицы n на n
def gen_matrix(n):
    fmatrix = []
    current_num = 1 # вводим переменную которая потом будет последовательно увеличиваться, чтобы создать матрицу с натуральными числами
    for i in range(n):
        row = []
        for j in range(n):
            row.append(current_num)
            current_num += 1
        fmatrix.append(row)
    return(fmatrix)

n = int(input("Введите размер матрицы n: "))
matrix = gen_matrix(n)

# Шаг 2: Подсчет сумм

# Суммы строк 
row_sums = []
for r in matrix:
    row_sums.append(sum(r))

# Суммы столбцов
col_sums = []
for j in range(n):
    s = 0
    for i in range(n):
        s += matrix[i][j]
    col_sums.append(s)

# Суммы диагоналей
sum1 = 0 # сумма одной диагонали 
sum2 = 0
duo_sum = 0 # сумма двух диагоналей

# Подсчет sum1
for i in range(n):
    sum1 += matrix[i][i]

# Подсчет sum2
for i in range(n):
    sum2 += matrix[n-1-i][i]

# Подсчет duo_sum 
if n % 2 == 0:
    duo_sum = sum1 + sum2
else:
    f = n // 2 
    duo_sum = sum1 + sum2 - matrix[f][f] #вычитаем средний элемент матрицы

# Шаг 3: Проверка условий и выводим результат

all_sums = row_sums + col_sums # Объединяем все суммы строк и столбцов в один список для удобства проверки

# Проверка первого условия
check1 = 0
for r in all_sums:
    if r == sum1:
        check1 += 1
if check1 > 0: print(f'Сумма одной диагонали {sum1}')

# Проверка второго условия
check2 = 0
for r in all_sums:
    if r == duo_sum:
        check2 += 1
if check2 > 0: print(f'Сумма одной диагонали {sum1}, а во второй {sum2}')

# Проверка третьего условия
if check1 + check2 == 0: print('Сумм нет')


# Задание 2

def check(int_log): 
    if int_log in ['Мавпродош', 'Лорнектиф', 'Древерол', 'Фиригарпиг', 'Клодобродыч']: # проверка наличия логина в данном списке
        print('есть логин с таким именем')
    else:
        while int_log not in ['Мавпродош', 'Лорнектиф', 'Древерол', 'Фиригарпиг', 'Клодобродыч']: # цикл, который продолжает попытку ввода, пока логин не совпадет с одним из списка
            int_log = input('Нет логина с таким именем, введите логин повторно: ')
        print('есть логин с таким именем')

int_log = input()
check(int_log)
````

### Задание 4 Графики

1. Необходимо написать функции вычисления

$f1 = 2x^2 + 6x + b$ -- синий

$x^2 + (f2 - \sqrt{|x|})^2 = 1$ - красный

$b$ - последняя цифра в студенческом билете, если она 0, то первая не 0 цифра в номере.

2. Построить графики функции:
   1. Одно окно и два графика (subplot) 
   2. Оба графика в одном окне на одном графике в разных координатных осях (ось OY у одного графика слева, а у второго справа)
   3. Подписать все оси и название графика и легенду.

````python
# импортируем библиотеки 
import numpy as np
import matplotlib.pyplot as plt

# Первое число студенческого билета
b = 9

# Определяем функции
# График синей функции 
def f1(x):
    return 2 * x**2 + 6 * x + b
# Красная функция задана неявно
# Поэтому разделим ее на две, где отличаются лишь знаки
def f2_upper(x):
    return np.sqrt(np.abs(x)) + np.sqrt(1 - x**2)

def f2_lower(x):
    return np.sqrt(np.abs(x)) - np.sqrt(1 - x**2)


# Создаём массивы аргументов
x_blue = np.linspace(-3, 3, 500)
x_red = np.linspace(-1, 1, 500)

# Создаём окно и два графика
plt.figure(figsize=(12, 5))

# Первый график, на котором синяя функция
plt.subplot(1, 2, 1)
# Настройки синего графика 
plt.plot(
    x_blue,
    f1(x_blue),
    color='blue',
    label='f1(x) = 2x² + 6x + b'
)
# Название для графика
plt.title('Синяя функция')
# Ниже подписываю оси
plt.xlabel('x')
plt.ylabel('y')
# pli.grid нужен для создание клеточек на заднем фоне графика
plt.grid()
plt.legend()

# Второй график, на котором красная функция
plt.subplot(1, 2, 2)

# Настройки красного графика
plt.plot(
    x_red,
    f2_upper(x_red),
    color='red',
    label='f2 верхняя ветвь'
)

plt.plot(
    x_red,
    f2_lower(x_red),
    color='red',
    label='f2 нижняя ветвь'
)

plt.title('Красная функция')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.legend()


# Автоматическая настройка отступов
plt.tight_layout()

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

# Первое число студенческого билета 
b = 9

# Функции
# График синей функции
def f1(x):
    return 2 * x**2 + 6 * x + b

def f2_upper(x):
    return np.sqrt(np.abs(x)) + np.sqrt(1 - x**2)

def f2_lower(x):
    return np.sqrt(np.abs(x)) - np.sqrt(1 - x**2)

# Диапазон для синей функции, где первые два параметра это диапозон, а последнее число это количество точек в нем
x_blue = np.linspace(-1, 1, 500)

# Диапазон для красной функции
x_red = np.linspace(-1, 1, 500)

# Создаём основную ось
fig, ax1 = plt.subplots(figsize=(10, 6))

# Синяя функция левая ось Y
ax1.plot(
    x_blue,
    f1(x_blue),
    color='blue',
    label='f1(x)'
)
# даем название оси
ax1.set_xlabel('x')
ax1.set_ylabel('f1(x)', color='blue')

# красим числа на оси OY в синий цвет
ax1.tick_params(axis='y', labelcolor='blue')

# Создаем клеточки на заднем фоне
ax1.grid()

# Создаём вторую ось Y для графика красного цвета, со своими делениями и цветом
ax2 = ax1.twinx()

# Красная функция  правая ось Y
ax2.plot(
    x_red, 
    f2_upper(x_red), # формула данной части графика
    color='red', # задаем цвет
    label='f2 верхняя ветвь' # подписываем для легенды графика 
)

ax2.plot(
    x_red,
    f2_lower(x_red),
    color='red',
    linestyle='--', # чтобы линиия была в виде пунктиров
    label='f2 нижняя ветвь'
)

ax2.set_ylabel('f2(x)', color='red')

ax2.tick_params(axis='y', labelcolor='red')


# Название графика
plt.title('Функции f1 и f2 на графике')

# Объединяем легенды обеих осей
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    lines1 + lines2,
    labels1 + labels2,
    loc='upper left' # Где будет находится легенда графика
)

plt.tight_layout()
plt.show()
````
