## Задание 3

Написать функцию расчета действительных корней квадратного уравнения вида: $$ax^2+bx+c=0$$
Функция запрашивает вводом с клавиатуры действительные коэффициенты a, b и с. 
Функция выводит действительные корни уравнения $x_1$ и  $x_2$:
```
Корни уравнения
x1 =
x2 =
```

или

```
Корни уравнения
x1 = x2 =
```

или

```
Действительных корней нет
```

или

```
Действительный корень x = 
```

````python
import numpy as np

def quadratic():
    print('Решение уравнения ax**2 + bx + c = 0')

    # Ввод значений коэффициентов
    a = float(input('Введите коэффициент a: '))
    b = float(input('Введите коэффициент b: '))
    c = float(input('Введите коэффициент c: '))

    # Проверка частных случаев
    if a == 0:
        if b == 0:
            if c == 0:
                print('Уравнение верно для любого x')
            else:
                print('Действительных корней нет')
        else:
            x = -c / b
            print(f'Действительный корень x = {x}')
        return

    # Основной алгоритм для квадратного уравнения
    D = b**2 - 4 * a * c
    print(f'Дискриминант D = {D}')

    if D < 0:
        print('Действительных корней нет')

    elif D == 0:
        x = -b / (2 * a)
        print('Корни уравнения')
        print(f'x1 = x2 = {x}')

    else:
        x1 = (-b + np.sqrt(D)) / (2 * a)
        x2 = (-b - np.sqrt(D)) / (2 * a)
        print('Корни уравнения')
        print(f'x1 = {x1}')
        print(f'x2 = {x2}')

quadratic()
````

## Задание 4

Функция получает на вход действительное число $h$. Вычисляет корни уравнения $ax^2+bx+c=0$ действительные корни, если
$$a = \sqrt{\frac{|\sin(8h)|+17}{(1-\sin(4h)\cos(h^2+18))^2}}$$
$$b = 1 - \sqrt{\frac{3}{3+|\tan(ah^2)-\sin(ah)|}}$$
$$ с=ah^2\sin(bh)+bh3\cos(ah)$$

Если корни существуют, то выводит их, иначе выводит сообщение, что корней нет.


````python
import math

def f(h):
    a = math.sqrt(
        (abs(math.sin(8 * h)) + 17)
        / (1 - math.sin(4 * h) * math.cos(h**2 + 18))**2
    )

    b = 1 - math.sqrt(
        3 / (3 + abs(math.tan(a * h**2) - math.sin(a * h)))
    )

    c = (
        a * h**2 * math.sin(b * h)
        + b * h**3 * math.cos(a * h)
    )

    D = b**2 - 4 * a * c

    if D < 0:
        return "Действительных корней нет"

    if D == 0:
        x = -b / (2 * a)
        return x

    x1 = (-b + math.sqrt(D)) / (2 * a)
    x2 = (-b - math.sqrt(D)) / (2 * a)

    return x1, x2

h = float(input("Введите h: "))
print(f(h))
````

## Задание 8

Начав тренировки, лыжник в первый день пробегал 10 км. Каждый следующий день он увеличивал длину пробега на 10% от предыдущего дня. Определить, в какой день он пробежит больше 20 км, в какой день суммарный пробег за все дни превысит 100 км.

Написать функцию, которая на вход получает количество км, а выводит через сколько дней от начала тренировок лыжник суммарно за все дни пробежит больше введенного значения.

````python
def skiing(km):
    day = 1
    distance = 10
    total = 10

    day_20 = 0

    while distance <= 20:
        day += 1
        distance *= 1.1

    day_20 = day

    day = 1
    distance = 10
    total = 10

    while total <= km:
        day += 1
        distance *= 1.1
        total += distance

    return day_20, day


km = float(input("Введите количество километров: "))

day_20, total_day = skiing(km)

print("Больше 20 км лыжник пробежит в", day_20, "день")
print("Суммарный пробег превысит", km, "км в", total_day, "день")
````

## Задание 9

Написать функцию, которая на вход принимает двумерный массив и номер строки сортировки, а выводит двумерный массив, у которого элементы указанной строки упорядочены по убыванию. Номер управляющей строки вводить с клавиатуры. 

Например, пусть исходный массив имеет вид:
$$\begin{matrix}
1 & 23 & 56 & 7 \\ 
44 & 5 & 38 & 9 \\ 
2 & 78 & 23 & 5 
\end{matrix}$$
номер управляющей строки – 3, тогда результирующий массив будет вы-глядеть так:
$$\begin{matrix}
1 & 23 & 56 & 7 \\ 
44 & 5 & 38 & 9 \\ 
2 & 5 & 23 & 78 
\end{matrix}$$


````python
def sort_row(matrix, row_number):
    row_index = row_number - 1

    matrix[row_index].sort(reverse=True)

    return matrix


matrix = [
    [1, 23, 56, 7],
    [44, 5, 38, 9],
    [2, 78, 23, 5]
]

row_number = int(input("Введите номер строки: "))

result = sort_row(matrix, row_number)

for row in result:
    print(row)
````

## Задание 10

Написать функцию, которая получает на вход значения $a$ и $b$ и вычисляет интеграл

$$\int_{a}^{b}\cos e^xdx$$

методом прямоугольников при n=100 (n – число интервалов разбиения)

````python
import math

a = float(input("Введите a: "))
b = float(input("Введите b: "))

def integral(a, b):
    n = 100
    h = (b - a) / n
    result = 0

    for i in range(n):
        x = a + i * h
        result += math.cos(math.exp(x))

    result *= h

    return result

print("Значение интеграла:", integral(a, b))
````
