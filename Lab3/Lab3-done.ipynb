## Теория

В классической механике для описания механических явлений, в качестве первого шага необходимо выбрать ту или иную систему отсчета. В различных системах отсчета законы движения имеют различный вид. Для нахождения траекторий относительных движений в классической механике используется предположение об абсолютности времени во всех системах отсчета (как в инерциальных, так и в неинерциальных).

Рассмотрим движение одной и той же точки в двух различных системах отсчета $K$ и $K'$, при этом вторая движется относительно первой со скоростью $V(t)=\frac{d}{dt}R(t)$, где $R(t)$ - радиус вектор, описывающий положение точки начала системы координат $K'$ относительно системы отсчета $K'$.

Движение рассматриваемой точки относительно системы отсчета $K$ описывается радиусом-вектором $r(t) = r'(t) + R(t)$, здесь $r(t)$ - это радиус вектор в системе $K'$. Оросительная скорость может быть рассчитана согласно $v(t)=\frac{d}{dt}r(t)=\frac{d}{dt}r'(t)+\frac{d}{dt}R(t)$. В формуле $\frac{d}{dt}r'(t)$ - скорость точки относительно системы отсчета $K'$, $\frac{d}{dt}R(t)$ - скорость движения системы отсчета $K'$ относительно системы отсчета $K$.

Для нахождения закона движения точки в произвольной системе отсчета $K$ необходимо:
1. Задать закон движения точки относительно системы отсчета $K'$ - функцию $r'(t)).
2. Задать закон движения системы $K'$ относительно системы отсчета $K$ - функцию $R(t)$.
3. Определить закон движения точки относительно системы отсчета $K$ - выражение $r(t) = r'(t) + R(t)$.

## Построение орбиты Луны в гелиоцентрической системе отсчета

С точки зрения кинематического подхода данная задача соответствует первой постановке задачи об относительном движении. В гелиоцентрической системе отсчета (СО) (СО $K$) Земля движется по окружности радиуса $R_1=1.496 * 10^8$ км (период обращения $T_1=3.156 * 10^7$ с). Луна движется вокруг Земли (СО $K$) по окружности радиуса $R_2 = 3.844 * 10^5$ км (период обращения $T_2 = 2.36 * 10^6$ с). При движении материальной точки по окружности радиуса $R$ с постоянной угловой скоростью $\omega$ координаты радиуса-вектора, проведенного из начала координат к текущему положению точки меняются по закону $R(t)$:

$R \cos(\omega t + \phi_0) = R \cos( \frac{2 \pi}{T}t+ \phi_0)$

$R \sin(\omega t + \phi_0) = R \sin( \frac{2 \pi}{T}t+ \phi_0)$

где $\phi_0$ - начальная фаза, характеризующая положение частицы в момент времени $t=0$, которую в данной работе можно принять равной 0. Заменив в уравнении выше $R$ на $R_1,R_2$ и подставив в $r(t) = r'(t) + R(t)$, будет получена зависимость радиуса-вектора Лены в гелиоцентрической системе координат от времени $r(t)$:

$x(t) = R_2 \cos(\frac{2 \pi}{T_2}t) + R_1 \cos(\frac{2 \pi}{T_1}t)$

$y(t) = R_2 \sin(\frac{2 \pi}{T_2}t) + R_1 \sin(\frac{2 \pi}{T_1}t)$

Данное выражение задает орбиту Луны $y=y(x(t))$ в параметрической форме, нде параметром является время.

Исходные данные:

* $R_1 = 1.496 * 10^8$ - численное значение радиуса орбиты Земли
* $T_1 = 3.156 * 10^7$ - численное значение периода обращения Земли вокруг Солнца
* $R_2 = 3.844 * 10^5$ - численное значение радиуса орбиты Луны
* $T_2 = 2.360 * 10^6$ - численное значение периода обращения Земли вокруг своей оси

```python
# Исходные значения из условия: расстояния в км, периоды в секундах.
R1 = 1.496 * 10**8  # Радиус орбиты Земли.
T1 = 3.156 * 10**7  # Период обращения Земли вокруг Солнца.
R2 = 3.844 * 10**5  # Радиус орбиты Луны вокруг Земли.
T2 = 2.360 * 10**6  # Период обращения Луны вокруг Земли.
```

Задайте ниже переменную времени t от 0 до T1 с шагом T1 \ 1000

```python
# 1000 равных интервалов на протяжении одного земного года.
t = np.arange(0, T1 + T1 / 2000, T1 / 1000)
```

Как видно данные орбиты имеют разную форум, для понимания причин отличия формы необходимо сравнить линейные скорости движения Луны в обоих случаях и линейную скорость движения Земли. Так как направление линейной скорости движения Земли относительно Солнца и направления линейной скорости движения Луны относительно Земли меняются во времени и при этом по величине эти скорости остаются постоянными, в качестве количественной характеристики соотношения линейных скоростей движения Луны и Земли в гелиоцентрической системе координат следует выбирать разность между модулем линейной скорости движения Земли и проекцией линейной скорости Луны на направление вектора линейной скорости Земли:

$v(t) = |V(t)| - \frac{V(t)v(t)}{|V(t)|}$,

где $V(t)$ - вектор скорости движения Земли относительно Солнца

$\frac{dX(t)}{dt}$

$\frac{dY(t)}{dt}$

$v(t)$ - вектор скорости Луны относительно Земли


$\frac{dx(t)}{dt}$

$\frac{dy(t)}{dt}$

### Выполните вычисление скоростей Земли и Луны для времени с шагом $dt = \frac{T1}{2000}$ согласно теории и описанию в комментариях, заполняя пропущенные строки кода. Отобразите полученный результат.

```python
# Задаем исходные данные (км и с).
R1 = 1.496 * 10**8  # Радиус земной орбиты.
T1 = 3.156 * 10**7  # Земной год.
R2 = 3.844 * 10**5  # Радиус лунной орбиты.
T2 = 2.360 * 10**6  # Лунный период.

# Используем вдвое более мелкий шаг времени, чем в первом задании.
dt = T1 / 2000
t = np.arange(0, T1 + dt / 2, dt)

# Положение Земли и Луны в соответствующих системах отсчета.
Xe = R1 * np.cos(2 * np.pi * t / T1)  # Земля: X.
Ye = R1 * np.sin(2 * np.pi * t / T1)  # Земля: Y.
Xm = R2 * np.cos(2 * np.pi * t / T2)  # Луна относительно Земли: X.
Ym = R2 * np.sin(2 * np.pi * t / T2)  # Луна относительно Земли: Y.
Xrelative = Xe + Xm  # Луна относительно Солнца: X.
Yrelative = Ye + Ym  # Луна относительно Солнца: Y.

# Скорости рассчитываем численным дифференцированием координат.
Vx = np.gradient(Xe, dt)  # Скорость Земли вдоль X, км/с.
Vy = np.gradient(Ye, dt)  # Скорость Земли вдоль Y, км/с.
vx = np.gradient(Xm, dt)  # Скорость Луны относительно Земли вдоль X.
vy = np.gradient(Ym, dt)  # Скорость Луны относительно Земли вдоль Y.

# |V_земли| - проекция скорости Луны на направление скорости Земли.
Vearth = np.hypot(Vx, Vy)
V = Vearth - (Vx * vx + Vy * vy) / Vearth

# Согласуем время и численно полученные скорости.
tplot = t / 86400  # Переводим секунды в дни для подписей.

f = plt.figure(figsize=(12, 4))
ax = f.add_subplot(121)
ax2 = f.add_subplot(122)
ax.plot(Xe, Ye, 'b', label='Земля')
ax.plot(Xrelative, Yrelative, 'r', label='Луна')
ax.set_title('Орбиты относительно Солнца')
ax.set_xlabel('X, км')
ax.set_ylabel('Y, км')
ax.axis('equal')
ax.grid(True)
ax.legend()
ax2.plot(tplot, V)
ax2.set_title('Разность скорости Земли и проекции скорости Луны')
ax2.set_xlabel('Время, дни')
ax2.set_ylabel('v(t), км/с')
ax2.grid(True)
plt.tight_layout()
plt.show()
```

## Построение орбиты Марса в системе отсчета, связанной с Землей

С точки зрения кинематического подхода данная задача соответствует второй постановке задачи об относительном движении. В гелиоцентрической системе отсчета (СО) (СО $K$) Земля движется по окружности радиуса $R_1=1.496 * 10^8$ км (период обращения $T_1=365.2$ суток). Марс движется по эллипсу, большая полуплоскость которого $a_M=2.2 * 10^8$ км, период обращения $T_M = 689.98$ суток, эксцентриситет орбиты = 0.093. Движение Земли описывается радиус-вектором $R(t)$, а так как орбита Макса эллипс зависимости $x=x(t), y=y(t)$ от времени задаются параметрически как. 

$x(\xi) = a_M(\cos\xi-e)$,

$y(\xi) = a_m \sqrt {1-e} \sin\xi$, 

$t(\xi) = \frac{T_M}{2\pi}(\xi-e\sin\xi)$.

Полному обороту по эллипсу соответствует изменение параметра $\xi$ от 0 до $2\pi$.

Для построения орбиты Марса необходимо вычислить в одни и те же моменты времени координаты радиусов-векторов, описывающих положение Земли и Марса в гелиоцентрической системе отсчета, затем согласно $r'(t)=r(t)-R(t)$ вычислить координаты Марса в системе отсчета, связанной с Землей.

Алгоритм:

1. Задание интервала изменения параметра $\xi$.
2. Разбиение интервала изменения параметра $\xi$ на $N$ последовательных интервалов.
3. Вычисление для каждого значения $\xi$ значений времени и координат $x,y$ согласно $x(\xi) = a_M(\cos\xi-e)$,  $y(\xi) = a_m \sqrt {1-e} \sin\xi$, $t(\xi) = \frac{T_M}{2\pi}(\xi-e\sin\xi)$.

### Выполните вычисление координат Земли и Марса в гелиоцентрической системе координат согласно теории и описанию в комментариях, заполняя пропущенные строки кода. Отобразите полученный результат.

```python
# Радиусы в км, периоды обращения в сутках.
R1 = 1.496 * 10**8  # Радиус орбиты Земли.
T1 = 365.2  # Период Земли.
Am = 2.2 * 10**8  # Большая полуось орбиты Марса.
Tm = 689.98  # Период Марса.
E = 0.093  # Эксцентриситет Марса.
Np = 1000  # Точек на один оборот Марса.

dksi = 2 * np.pi / Np  # Шаг эксцентрической аномалии.
ksi = np.arange(0, 2 * np.pi + dksi / 2, dksi)

# Из уравнения Кеплера находим момент времени для каждого ksi.
T = Tm / (2 * np.pi) * (ksi - E * np.sin(ksi))
Xm = Am * (np.cos(ksi) - E)  # Координата X Марса.
# Для эллипса малая полуось b = a * sqrt(1 - e²).
Ym = Am * np.sqrt(1 - E**2) * np.sin(ksi)  # Координата Y Марса.

# Положение Земли вычисляем в те же моменты времени T.
Xe = R1 * np.cos(2 * np.pi * T / T1)  # X Земли.
Ye = R1 * np.sin(2 * np.pi * T / T1)  # Y Земли.

fig = plt.figure(figsize=(8, 4), dpi=100)
plt.plot(Xe, Ye, 'r', label='Земля')
plt.plot(Xm, Ym, 'b', label='Марс')
plt.title('Орбиты Земли и Марса относительно Солнца')
plt.xlabel('X, км')
plt.ylabel('Y, км')
plt.axis('equal')
plt.grid(True)
plt.legend()
plt.show()
```

### Выполните вычисление координат Земли и Марса в системе координат относительно Земли согласно теории и описанию в комментариях, заполняя пропущенные строки кода. Отобразите полученный результат.

```python
# Моделируем несколько оборотов Марса для наблюдения петель.
R1 = 1.496 * 10**8  # Радиус орбиты Земли, км.
T1 = 365.2  # Период Земли, сутки.
Am = 2.2 * 10**8  # Большая полуось Марса, км.
Tm = 689.98  # Период Марса, сутки.
E = 0.093  # Эксцентриситет орбиты Марса.
K = 3  # Число полных оборотов Марса.
Np = 1000  # Число точек на один оборот.

dksi = 2 * np.pi / Np
ksi = np.arange(0, K * 2 * np.pi + dksi / 2, dksi)

# Параметризация эллиптической орбиты и закон времени Кеплера.
T = Tm / (2 * np.pi) * (ksi - E * np.sin(ksi))
Xm = Am * (np.cos(ksi) - E)
Ym = Am * np.sqrt(1 - E**2) * np.sin(ksi)

# Земля на круговой орбите в те же моменты T.
Xe = R1 * np.cos(2 * np.pi * T / T1)
Ye = R1 * np.sin(2 * np.pi * T / T1)

# Положение Марса относительно Земли = Марс минус Земля.
Xrelative = Xm - Xe
Yrelative = Ym - Ye

fig = plt.figure(figsize=(8, 5), dpi=100)
plt.plot(Xrelative, Yrelative, 'r', label='Марс относительно Земли')
plt.title('Видимая геоцентрическая траектория Марса')
plt.xlabel('X, км')
plt.ylabel('Y, км')
plt.axis('equal')
plt.grid(True)
plt.legend()
plt.show()
```

### Выполните вычисление расстояний между Землей и Марсом согласно теории и комментариям в коде. Отобразите полученный результат.

```python
# Теорема Пифагора для расстояния между планетами в каждый момент T.
R = np.sqrt(Xrelative**2 + Yrelative**2)

fig = plt.figure(figsize=(8, 4), dpi=100)
plt.plot(T, R)
plt.title('Расстояние между Землей и Марсом')
plt.xlabel('Время, сутки')
plt.ylabel('Расстояние, км')
plt.grid(True)
plt.show()

# Численные экстремумы помогают сформулировать вывод по заданию.
print(f'Минимальное расстояние: {R.min():.3e} км')
print(f'Максимальное расстояние: {R.max():.3e} км')
```

## Выберете планету Солнечной системы и смоделируйте ее орбиту в системе координат, связанной с Землей согласно астрономическим данным. Рассчитайте расстояние от планеты до Земли.

* Меркурий
* Венера
* Юпитер
* Сатурн
* Уран
* Нептун

```python
# Выбранная планета — Юпитер. Приближенные астрономические данные:
# большая полуось 778.57 млн км, эксцентриситет 0.0489, период 4332.59 суток.
# Как и для Марса, используем параметризацию эллипса через ksi.
R_zemli = 1.496 * 10**8  # Радиус круговой орбиты Земли, км.
T_zemli = 365.2  # Период обращения Земли, сутки.
a_yupitera = 7.7857 * 10**8  # Большая полуось Юпитера, км.
T_yupitera = 4332.59  # Орбитальный период Юпитера, сутки.
e_yupitera = 0.0489  # Эксцентриситет орбиты Юпитера.
kolichestvo_oborotov = 2
kolichestvo_tochek = 2500

# По аналогии с задачей о Марсе разбиваем интервал параметра ksi.
ksi_y = np.linspace(0, 2 * np.pi * kolichestvo_oborotov,
                    kolichestvo_tochek * kolichestvo_oborotov + 1)
vremya_y = T_yupitera / (2 * np.pi) * (ksi_y - e_yupitera * np.sin(ksi_y))

# Гелиоцентрические координаты Юпитера.
x_y = a_yupitera * (np.cos(ksi_y) - e_yupitera)
y_y = a_yupitera * np.sqrt(1 - e_yupitera**2) * np.sin(ksi_y)

# Гелиоцентрические координаты Земли в те же моменты времени.
x_z = R_zemli * np.cos(2 * np.pi * vremya_y / T_zemli)
y_z = R_zemli * np.sin(2 * np.pi * vremya_y / T_zemli)

# Переход к системе отсчета, связанной с Землей.
x_otn = x_y - x_z
y_otn = y_y - y_z
rasstoyanie = np.hypot(x_otn, y_otn)  # Расстояние Земля–Юпитер, км.

# Показываем траекторию и изменение расстояния на двух графиках.
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(x_otn, y_otn, label='Юпитер относительно Земли')
axes[0].set_title('Геоцентрическая траектория Юпитера')
axes[0].set_xlabel('X, км')
axes[0].set_ylabel('Y, км')
axes[0].axis('equal')
axes[0].legend()
axes[0].grid(True)
axes[1].plot(vremya_y, rasstoyanie)
axes[1].set_title('Расстояние Земля — Юпитер')
axes[1].set_xlabel('Время, сутки')
axes[1].set_ylabel('Расстояние, км')
axes[1].grid(True)
plt.tight_layout()
plt.show()
print(f'Минимальное расстояние: {rasstoyanie.min():.3e} км')
print(f'Максимальное расстояние: {rasstoyanie.max():.3e} км')
```
